```python
import numpy as np

def zero_cross(ar_of_a_sampled_f):
    """Returns indices where the signal changes sign."""
    ar_of_sign_change_indxs = np.nonzero(np.diff(np.sign(ar_of_a_sampled_f)))[0]
    return ar_of_sign_change_indxs

v0 = float(input())

G = 9.8  # gravitational acceleration (m/s²)
t = np.arange(0, 5, 0.01)
y = -0.3 + (v0 * t) - (0.5 * G * (t ** 2))

# Find indices where height crosses zero (ground level)
crossing_indexes = zero_cross(y)

# Map those indices to actual timestamps
crossing_times = t[crossing_indexes]

# Time of flight = last ground crossing minus first ground crossing
time_of_flight = crossing_times[-1] - crossing_times[0]

#output
#First answer
print(f'time of flight: {time_of_flight:.5g} s')

#Second answer
#roots = np.roots([a,b,c])
#time_of_flight = roots[1] - roots[0]
#print(f'time of flight: {time_of_flight:.5g} s')
```
