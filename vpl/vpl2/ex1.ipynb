```python

import numpy as np
import numpy.typing as npt

def zero_cross(ar):
    #returns an array of indices where ar changes sign
    H1t = np.sign(ar)
    H1s = np.abs(H1t[:-1] - H1t[1:])
    return np.nonzero(H1s)[0]


#input
T1 = float(input('enter T_one: '))
T2 = float(input('enter T_two: '))
qn = int(input('enter question no.: '))

R = 40


t: npt.NDArray[np.float64]

w1 = 2 * np.pi / T1
w2 = 2 * np.pi / T2


# ex 1
def calculate_vec_x_vec_y(R, t, w):
    o = w * t
    vec_x, vec_y = R * np.cos(o), R * np.sin(o)
    return vec_x, vec_y

#output
if qn==1:
    t = np.linspace(0, T1/6, 300)
    vec_x , vec_y = calculate_vec_x_vec_y(R, t,w1)
    X1 = vec_x[-1]
    Y1 = vec_y[-1]
    print(f'{X1:.5g},{Y1:.5g} m')
if qn==2:
    t = np.linspace(0, T1/6, 300)
    vec_x_1 , vec_y_1 = calculate_vec_x_vec_y(R, t,w1)
    vec_x_2 , vec_y_2 = calculate_vec_x_vec_y(R, t,w2)
    d = np.sqrt((vec_x_2[-1] - vec_x_1[-1])**2 + (vec_y_2[-1] - vec_y_1[-1])**2)
    print(f'{d:.5g} m')
if qn==3:
    t = np.linspace(0, T1/6, 300)
    o1 = (w1 * t)[-1]
    o2 = (w2 * t)[-1]
    s = R * abs(o2 - o1)
    print(f'{s:.5g} m')
if qn == 4:
    t = np.linspace(0, T2 * 10, 10_000)
    rel = np.sin((w1 - w2) * t)
    crossings = zero_cross(rel)
    upward = crossings[rel[crossings] < 0]  # sin: - to + means relative angle = 2πn (meeting)
    T3_numeric = t[upward[0]]
    print(f'{T3_numeric:.5g}')
if qn==5:
    T3_analytic = (T1 * T2) / abs(T1 - T2)
    print(f'{T3_analytic:.5g}')


```
